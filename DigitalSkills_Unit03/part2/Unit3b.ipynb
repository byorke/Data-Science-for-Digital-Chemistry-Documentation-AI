## Common Error Messages ##
Here is a list of common error messages from the packages we have used and what they mean. You can also download a pdf version of this list <a href='datasets/common_errors.pdf'>Common errors</a>

## 🐍 Common Python Error Messages
### Core Python Errors
1. SyntaxError: invalid syntax
Occurs when Python encounters code it cannot parse due to typos, missing punctuation, wrong indentation, or mismatched parentheses.
2. IndentationError: unexpected indent
IndentationError: unindent does not match any outer indentation level
Happens when indentation is inconsistent (mix of tabs/spaces or misaligned blocks).
3. NameError: name 'x' is not defined
Raised when using a variable or function before it is defined, or if it is misspelled.
4. TypeError: unsupported operand type(s)
Occurs when an operation is applied to incompatible data types.
5. ValueError: invalid literal for int()
Raised when a function receives the right type but an inappropriate value.
6. AttributeError: 'object' has no attribute 'x'
You attempted to access an attribute or method that does not exist.
7. IndexError: list index out of range
You attempted to access a list index that doesn’t exist.
8. KeyError: 'some_key'
Occurs when accessing a dictionary key that isn’t present.
9. ImportError: cannot import name X
ModuleNotFoundError: No module named 'X'
Occurs when a module or object is not installed, not available, or misspelled.
10. ZeroDivisionError: division by zero
Raised when dividing any number by zero.

### 🔢 NumPy Error Messages
1. ValueError: operands could not be broadcast together
Occurs when performing operations on arrays with incompatible shapes.
2. TypeError: ufunc 'add' did not contain a loop with signature...
NumPy’s ufunc cannot operate on the given datatype(s).
3. IndexError: too many indices for array
Indexing uses more dimensions than the array actually has.
4. ValueError: cannot reshape array of size X into shape Y
Reshape fails because the number of elements does not match.
5. **RuntimeWarning: overflow encountered in ... or RuntimeWarning: divide by zero encountered in ...**
Warnings for numerical instability.

### 🧮 Pandas Error Messages
1. KeyError: 'column_name'
The most common Pandas error. A column label does not exist.
2. IndexingError: Too many indexers
Occurs when indexing with too many dimensions.
3. ValueError: Length of values does not match length of index
Happens when assigning data whose length doesn’t match the DataFrame index.
4. SettingWithCopyWarning
A warning indicating ambiguous chained indexing (copy vs view).
5. TypeError: cannot concatenate object of type ...
Raised during concatenation of incompatible objects.
6. ParserError (when reading CSVs)
Occurs when the CSV file is badly formatted.

### 📊 Matplotlib Error Messages
1. ValueError: x and y must have same first dimension
Plotting arrays of different lengths.
2. TypeError: float() argument must be a string or a number
Matplotlib receives non‑numeric input.
3. AttributeError: 'AxesSubplot' object has no attribute ...
Using the wrong API (mixing plt and object‑oriented approaches).
4. RuntimeError: Invalid DISPLAY variable
Triggered in headless environments lacking a GUI backend.
5. UserWarning: Glyph X missing from current font
Occurs when Matplotlib cannot render specific Unicode characters.

# More advanced debugging.

When you have a block of code sometimes the part causing the error is not the actual line of code where the error is occurring but something you have done before and this is where the fix needs to be made.  A useful strategy is to temporarily deactivate parts of the code by 'commenting out' the block of text.  By turning a block of code into a comment, it is not executed so you can execute bits of the code anc check if things are working as you expect.  By commenting the code rather than deleting it you don't have to rewrite large blocks of code if they turned out to be OK!  Remember you can quickly toggle whether a line of code is commented ( has a # at the start) by pressing the <kbd>Ctrl</kbd> and <kbd>/</kbd> keys. It is normally good practice to then inspect your data at that point and try to work out what has happened.  Can you work out where the issue is in the code below using these tools and fix the code?

```python
# create some simulated experimental data with noise
# create 500 points of x data between 0 and 10
x = np.linspace(0, 10, 500)

#create some random noise
noise = np.random.normal(0, 0.1, (500, 1))   
# use the noise and maths to create a simulated sin output
y = np.sin(x) + noise                        

#plot the data
fig, ax = plt.subplots()
ax.plot(x, y)  
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Example of a noisy sine wave")
plt.show()
```
