<div class="alert alert-success"><b>Task 1: Unit Conversion Functions.</b>
Please modify the code below to write a function called Calculate_in_kjmol. This should take in a list of energy values in kcal/mol (kilocalorie is a unit of energy often used to measure energy of food and while kcal/mol is commonly used in the US as a unit of energy in chemistry) and return a list, or array, of the same length where each value has been converted to kJ/mol. You perform this conversion by multiplying the value in kcal/mol by the conversion factor 4.184.

As an example if you ran the command 

>```python
>New_List = Calculate_in_kjmol(Example_list)
>```

You would expect the new list to look like

>```python
> [136.3984, 189.1168, 241.8352, 93.7216, 258.9896]
>```

Where each value in the original list has been converted from kcal/mol to kJ/mol

````python
Example_list = [32.6, 45.2, 57.8, 22.4, 61.9]


def Calculate_in_kjmol(energy_in_kcal):
    conversion = 4.184   # cm /s
    values = []
    for i in energy_in_kcal:
        values.append(i*conversion)
    return values

# alternatively use numpy, avoids the need for a for loop and is significantly quicker, particularly for long lists

def Calculate_in_kjmol_numpy(energy_in_kcal):
    conversion = 4.184
    import numpy as np
    values = np.array(energy_in_kcal)* conversion
    return values


````
